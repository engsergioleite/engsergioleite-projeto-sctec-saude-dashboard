```python
import pandas as pd
```

```python
df_2020 = pd.read_csv('../data/raw/2020.csv', sep=';')
df_2020.head()
```

```python
print(df_2020.shape)
df_2020.columns.tolist()
```

```python
anos = range(2020, 2027)
dfs = {}

for ano in anos:
    caminho = f'../data/raw/{ano}.csv'
    try:
        df = pd.read_csv(caminho, sep=';')
        dfs[ano] = df
        print(f"{ano}: OK - shape={df.shape}")
    except Exception as e:
        print(f"{ano}: ERRO - {type(e).__name__} - {e}")
```

```python
colunas_2020 = set(dfs[2020].columns)

for ano, df in dfs.items():
    cols = set(df.columns)
    faltando = colunas_2020 - cols
    extras = cols - colunas_2020
    if faltando or extras:
        print(f"{ano}: faltando={faltando} | extras={extras}")
    else:
        print(f"{ano}: colunas identicas a 2020")
```
